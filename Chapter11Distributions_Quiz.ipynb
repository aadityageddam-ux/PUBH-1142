# Chapter 11 Distributions - Quiz Questions

Use the random seed value of **12** to answer the following questions.

```python
import numpy
from plotly import graph_objects, figure_factory
import pandas
from scipy import stats
```

```python
# Population setup (seed 28, normal distribution mean=85, std=10, n=100000)
numpy.random.seed(28)
dBP = numpy.random.normal(85, 10, 100000)
```

```python
# Calculate the population mean
dBP.mean()
```

## Quiz Questions (use random seed 12)

```python
# Use random seed 12
numpy.random.seed(12)

# 1. Select a random sample of 40 individuals from the population
sample = numpy.random.choice(dBP, 40)

# 2. Calculate the sample mean
sample_mean = sample.mean()
print('sample_mean:', sample_mean)

# 3. Calculate the standard deviation of the sample
sample_std = sample.std(ddof=1)
print('sample_std:', sample_std)

# 4. Calculate the standard error
standard_error = sample_std / numpy.sqrt(40)
print('standard_error:', standard_error)

# 5. Calculate the confidence coefficient for a 95% CI (df=39)
c = stats.t.ppf(0.975, 39)
print('c:', c)

# 6. Calculate the lower bound of the confidence interval
lower_bound = sample_mean - c * standard_error
print('lower_bound:', lower_bound)

# 7. Calculate the upper bound of the confidence interval
upper_bound = sample_mean + c * standard_error
print('upper_bound:', upper_bound)

# 8. Create 1000 bootstrapped sample means (size 40) sorted ascending
bootstrapped_means = [numpy.random.choice(sample, 40, replace=True).mean() for _ in range(1000)]
bootstrapped_means.sort()

# 9. Show bootstrapped value at 2.5th percentile (index 24)
print('2.5th percentile:', bootstrapped_means[24])

# 10. Show bootstrapped value at 97.5th percentile (index 974)
print('97.5th percentile:', bootstrapped_means[974])
```
