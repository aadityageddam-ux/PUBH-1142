## Package Imports

```python
import pandas
import seaborn
from matplotlib import pyplot

# Set high resolution figures
%config InlineBackend.figure_format = 'retina'

# Set seaborn style
seaborn.set_style(style='whitegrid')
```

## Data Import

```python
# Import the heart failure dataset
df = pandas.read_csv('https://raw.githubusercontent.com/juanklopper/TutorialData/main/heart_failure.csv')

# Display first few rows
df.head()
```

## Task 1: Scatter Plot - Platelets vs Serum Sodium

Create a scatter plot to visualize the correlation between `platelets` and `serum_sodium` variables, colored by the unique elements of the `diabetes` variable. Add an appropriate title and axes labels.

```python
seaborn.scatterplot(
    data=df,
    x='platelets',
    y='serum_sodium',
    hue='diabetes'
).set(
    title='Platelets and Serum Sodium',
    xlabel='Platelets [platelets/mL]',
    ylabel='Serum sodium [mEq/L]'
);
```

## Task 2: Horizontal Histogram of Platelets

Generate a histogram to visualize the distribution of the `platelets` variable. Calculate the minimum and maximum values and use bins of 50,000. Use `y` instead of `x` to create a horizontal histogram with better spacing for tick values.

```python
# Check min and max values
print(f"Min platelets: {df.platelets.min()}")
print(f"Max platelets: {df.platelets.max()}")

# Create horizontal histogram
seaborn.displot(
    df,
    y='platelets',
    bins=range(0, 850000, 50000)
).set(
    title='Platelet Distribution',
    ylabel='Platelets [platelets/mL]',
    xlabel='Frequency'
);
```

## Task 3: Bar Plot - Diabetes with Death as Hue

Create a bar plot using the `diabetes` variable and add the `death` variable using the `hue` argument.

```python
seaborn.catplot(
    data=df,
    x='diabetes',
    hue='death',
    kind='count'
).set(
    title='Diabetes by Death Status',
    xlabel='Diabetes',
    ylabel='Frequency'
);
```
