### TASK 2 -  Calculate (6+4) × 2^10 ###

### Task 4 — x = 5, y = x, then x = 7 — What is the value of y? ###

```python
x = 5
y = x   # y is assigned the VALUE 5 (not a reference to x)
x = 7   # x now points to 7; y is unaffected

print(y)  # Output: 5
print(x)  # Output: 7
```

### Task 5 — For loop printing the square of each integer from 0 to 5 ###


```python
for i in range(6):   # range(6) gives 0, 1, 2, 3, 4, 5
    print(i**2)
```

### Task 6 — While loop: print square of i at each increment, starting at i = 0, while i < 10 ###


```python
i = 0                 # Initialize i to 0

while i < 10:         # Continue while i is less than 10
    print(i**2)       # Print the square of the current i
    i += 1            # Increment i by 1
```
