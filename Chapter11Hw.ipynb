## Exercise 1: Variables and Types

```python
# Create variables
age = 25
height = 5.9
first_name = "Alice"

# Print each variable and its type
print(f"age: {age}, type: {type(age)}")
print(f"height: {height}, type: {type(height)}")
print(f"first_name: {first_name}, type: {type(first_name)}")
```

## Exercise 2: Range Check

```python
# Check if n is between 50 and 100 (inclusive)
n = 75
result = 50 <= n <= 100
print(f"Is {n} between 50 and 100? {result}")

# Test with other values
n = 30
print(f"Is {n} between 50 and 100? {50 <= n <= 100}")

n = 100
print(f"Is {n} between 50 and 100? {50 <= n <= 100}")
```

## Exercise 3: String Operations

```python
s = "Python"

# String repetition
print(s * 5)

# String concatenation
print(s + " is fun!")
```

## Exercise 4: Type Conversion

```python
x = "12"
y = 8

# Convert x from string to integer to make the addition work
result = int(x) + y
print(result)
```
