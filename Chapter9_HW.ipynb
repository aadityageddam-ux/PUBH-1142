#### TASK 1 Determine the sample space elements, frequency, and relative frequencies of the diabetes variable ####

```python
# Sample space (unique values)
df.diabetes.unique()
# Output: array(['No', 'Yes'], dtype=object)

# Frequency count
df.diabetes.value_counts()
# Output:
# diabetes
# No    174
# Yes   125
# Name: count, dtype: int64

# Relative frequencies (proportions)
df.diabetes.value_counts(normalize=True)
# Output:
# diabetes
# No     0.581940
# Yes    0.418060
# Name: proportion, dtype: float64
```

#### The diabetes column has two classes: No and Yes. Out of 299 subjects, 174 (about 58.2%) did not have diabetes and 125 (about 41.8%) did. The mode of this variable is No.



### Task 2 — Generate a contingency table showing the joint percentages of the anaemia variable for each class in the death column


```python
pd.crosstab(df.death, df.anaemia, normalize=True) * 100
# Output:
# anaemia        No        Yes
# death
# No      40.133779  27.759197
# Yes     16.722408  15.384615
```

#### This cross-tab shows that among patients who survived (death = No), about 40.1% didn't have anaemia and 27.8% did. Among those who died (death = Yes), about 16.7% didn't have anaemia and 15.4% did.



### Task 3 — Determine the median ejection_fraction for each class in the anaemia variable


```python
df.groupby('anaemia')['ejection_fraction'].median()
# Output:
# anaemia
# No     38.0
# Yes    38.0
# Name: ejection_fraction, dtype: float64
```

#### Interestingly, the median ejection fraction is identical (38.0) for both patients with and without anaemia, suggesting that anaemia status alone doesn't seem to shift the central tendency of ejection fraction.
